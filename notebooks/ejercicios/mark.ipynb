Buscamos comparar 2 modelos de paneles solares. El Mark 1 que empieza a generar 1 kj/dia desde el dia 1 de abril. Mientras el Mark 2 que genera  4jk/dia, pero se pone en produccion el dia 1 de mayo es decir 30 dias mas tarde. Sabiendo esto buscamos saber que dia el Mark 2 alcanzara a Mark 1, cuanta energia habran generado ese dia y si Mark 2 genera 1 kj/dia en lugar de 4 cuando lo alcanza. 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0, 90, 1000) # inicio, final, n puntos
```

```python
d_r_m_1 = 1 * t
```

```python
d_r_m_2 = 4 * (t-30)
```

```python
fig, ax = plt.subplots()
plt.title('Comparativa Mark 1 y mark2')
plt.xlabel('Mark I')
plt.ylabel('Mark 2 ')
ax.set_xlim([0, 90])
ax.set_ylim([0, 120])
ax.plot(t, d_r_m_1, c='green')
ax.plot(t, d_r_m_2, c='brown')
plt.axvline(x=40, color='purple', linestyle='--')
_ = plt.axhline(y=40, color='purple', linestyle='--')
```

El mark 2 iguala al mark 1 a los 40 dias que seria el 10 de mayo

La energia generada es tan simple como solucionar la ecuacion en la que si resolvemos la de mark veremos que habra generado 40kj

```python
d_r_m_2_alternative = 1 * (t-30)
```

```python
fig, ax = plt.subplots()
plt.title('Comparativa Mark 1 y Mark2 alternativo')
plt.xlabel('Mark I')
plt.ylabel('Mark 2 ')
ax.set_xlim([0, 90])
ax.set_ylim([0, 120])
ax.plot(t, d_r_m_1, c='green')
ax.plot(t, d_r_m_2_alternative, c='brown')
```
