```python
import sympy as sp
```

```python
x = sp.symbols('x')
f = -5*x**3

df = sp.diff(f, x)      # f'(x) = -15x**2

```

```python
df
```

subs(x, valor) te da el valor de la pendiente (la derivada) en ese punto $x$ concreto — no el valor de la función original, sino el de su derivada $f'(x)=2x+2$ ya calculada.

```python
df.subs(x, 2)           # -60  -> confirma lo que sacaste con el método delta
```

```python
df.subs(x, -1)          # -15  -> confirma el ejercicio
```

```python
f = 2*x**2+3*2*x+2

df = sp.diff(f, x)      # f'(x) = 4x + 6

df
```

```python
df.subs(x, 2)           # 14  -> confirma lo que sacaste con el método delta
```

```python
df.subs(x, -1)          # 2   -> confirma el ejercicio
```
