```python
import numpy as np
```

```python
I = np.array([[1,0,0], [0,1,0], [0,0,1]])
u = np.array([[2,5,-3]]).T


np.dot(I, u)
```

```python
B = np.array([[2,0,-1], [-2,3,1], [0,4,-1]])


np.dot(B, u)
```
